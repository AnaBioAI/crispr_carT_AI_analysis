```python
import pandas as pd

genes = ["CD3", "CD4", "CD8"]

marker_category = {
    "CD3": "T-cell marker",
    "CD4": "Helper T-cell marker",
    "CD8": "Cytotoxic T-cell marker"
}

df = pd.DataFrame({"gene": genes})
df["category"] = df["gene"].map(marker_category)

print(df)
```

```python
df["category"].value_counts()
```
